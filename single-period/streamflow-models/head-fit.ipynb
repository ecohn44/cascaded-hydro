```python
def min_max(storage):
    return storage.min(), storage.max()
```

```python
def fit_head(S, H):
    mask = S.notna() & H.notna()
    S = S[mask]
    H = H[mask]

    a, b = np.polyfit(S, H, 1)
    err = H - (a * S + b)

    rmse = np.sqrt(np.mean(err**2))
    r2 = 1 - np.sum(err**2) / np.sum((H - H.mean())**2)

    print(f"h = {a:.2f} V_norm + {b:.2f}")
    print(f"RMSE: {rmse:.3f} m")
    print(f"R2: {r2:.3f}")

    return a, b
```

```python
BON["S_norm"] = normalize_forebay(BON, "bon")
```

```python
a, b = fit_head(BON["S_norm"], BON["head_m"])
```

# Reference Volume Trajectories

```python
volume_models = {}

# Normalized forebay is a proxy for normalized volume
for unit in forebay_limits:
    E_min, E_max = forebay_limits[unit]
    a = 1 / (E_max - E_min)
    b = -E_min / (E_max - E_min)
    volume_models[unit] = {"a": a, "b": b, "forebay_col": forebay_cols[unit]}
    print(f"{unit.upper()}: V_norm = {a:.4f} forebay + {b:.4f}")
```

```python
# These equations use fixed forebay limits and do not depend on the invalid cumulative storage series.
```

```python
def apply_fit(df, unit):
    
    # Get model fit
    m = volume_models[unit]

    # Get forebay 
    col = m["forebay_col"]

    # Compute daily average
    x = df.set_index("Date Time")[col].resample("D").mean().reset_index()

    # Get date 
    d = x["Date Time"]

    # Subset data to scenario date ranges 
    x = x[d.dt.year.between(2018, 2025) & d.dt.strftime("%m-%d").between("09-07", "12-05")].dropna().copy()

    # Calculate normalized volume reference
    x["V_norm"] = (m["a"] * x[col] + m["b"]).clip(0, 1)
    x["year"] = x["Date Time"].dt.year
    x["day"] = x.groupby("year").cumcount() + 1
    return x[["Date Time", "year", "day", "V_norm"]]

BON_ref = apply_fit(BON_daily, "bon")
TDA_ref = apply_fit(TDA_daily, "tda")
JDA_ref = apply_fit(JDA_daily, "jda")
MCN_ref = apply_fit(MCN_daily, "mcn")
```

```python
def apply_fit(df, unit):
    m = volume_models[unit]
    col = m["forebay_col"]

    x = df[["Date Time", col]].copy()
    x["Date Time"] = pd.to_datetime(x["Date Time"])
    x = x.drop_duplicates("Date Time").sort_values("Date Time")

    x["year"] = x["Date Time"].dt.year
    start = pd.to_datetime(x["year"].astype(str) + "-09-07")
    end = pd.to_datetime(x["year"].astype(str) + "-12-06")

    x = x[(x["year"].between(2018, 2025)) & (x["Date Time"] >= start) & (x["Date Time"] < end)].copy()
    x["V_norm"] = (m["a"]*x[col] + m["b"]).clip(0,1)
    x["hour"] = ((x["Date Time"] - start[x.index]).dt.total_seconds()/3600).astype(int) + 1

    return x[["Date Time", "year", "hour", "V_norm"]]

BON_ref = apply_fit(BON_H, "bon")
TDA_ref = apply_fit(TDA_H, "tda")
JDA_ref = apply_fit(JDA_H, "jda")
MCN_ref = apply_fit(MCN_H, "mcn")

refs = {"McNary": MCN_ref, "John Day": JDA_ref, "The Dalles": TDA_ref, "Bonneville": BON_ref}
fig, axes = plt.subplots(2, 2, figsize=(12,6), sharex=True, sharey=True)

for ax, (name, df) in zip(axes.flat, refs.items()):
    g = df.groupby("hour")["V_norm"]
    mean, p10, p90 = g.mean(), g.quantile(.1), g.quantile(.9)
    ax.fill_between(mean.index, p10, p90, alpha=.25, label="P10–P90")
    ax.plot(mean.index, mean, linewidth=2, label="Mean")
    ax.set(title=name, xlabel="Hour", ylabel="Normalized Storage", ylim=(0,1), xlim=(1,2160))

handles, labels = axes[0,0].get_legend_handles_labels()
fig.legend(handles, labels, loc="center right")
fig.tight_layout(rect=[0,0,.9,1])
plt.show()

refs = {"mcn": MCN_ref, "jda": JDA_ref, "tda": TDA_ref, "bon": BON_ref}
hourly_ref = None

for unit, df in refs.items():
    d = df[["Date Time", "V_norm"]].rename(columns={"V_norm": f"{unit}_soc"})
    hourly_ref = d if hourly_ref is None else hourly_ref.merge(d, on="Date Time", how="inner")

hourly_ref = hourly_ref.sort_values("Date Time").reset_index(drop=True)
hourly_ref = hourly_ref[["Date Time", "mcn_soc", "jda_soc", "tda_soc", "bon_soc"]]

print(hourly_ref.shape)
hourly_ref.head()
```

# Calculate Storage Range

```python
def cfs_to_kaf(cfs_day):
    """
    Convert 'CFS‑days' to acre‑feet:

    acre‑ft = cfs_day * 1.9835 / 1000
    """
    return cfs_day * 1.9835 / 1000
```

```python
## Storage Data (Change in CFS/day)
BON_S = pd.read_excel("/Users/elizacohn/Desktop/Columbia River Data/BPA Data/daily/daily/BON6S_daily.xlsx")
TDA_S = pd.read_excel("/Users/elizacohn/Desktop/Columbia River Data/BPA Data/daily/daily/TDA6S_daily.xlsx")
JDA_S = pd.read_excel("/Users/elizacohn/Desktop/Columbia River Data/BPA Data/daily/daily/JDA6S_daily.xlsx")
MCN_S = pd.read_excel("/Users/elizacohn/Desktop/Columbia River Data/BPA Data/daily/daily/MCN6S_daily.xlsx")

BON_S["date"] = pd.to_datetime(BON_S["date"]).dt.normalize()
TDA_S["date"] = pd.to_datetime(TDA_S["date"]).dt.normalize()
JDA_S["date"] = pd.to_datetime(JDA_S["date"]).dt.normalize()
MCN_S["date"] = pd.to_datetime(MCN_S["date"]).dt.normalize()
```

```python
import numpy as np
import pandas as pd

storage = {
    "bon": BON_S,
    "tda": TDA_S,
    "jda": JDA_S,
    "mcn": MCN_S,
}

Vrange_kaf = {}

for unit, df in storage.items():
    d = df.copy()
    d["date"] = pd.to_datetime(d["date"])

    # Keep Sep. 7–Dec. 5
    keep = (
        ((d["date"].dt.month == 9) & (d["date"].dt.day >= 7))
        | d["date"].dt.month.isin([10, 11])
        | ((d["date"].dt.month == 12) & (d["date"].dt.day <= 5))
    )

    d = d[keep].copy()
    d["year"] = d["date"].dt.year
    d["dS_kaf"] = d["S (unit:cfs)"] * 1.9835 / 1000

    seasonal_ranges = d.groupby("year")["dS_kaf"].apply(
        lambda x: np.ptp(np.r_[0, x.cumsum().to_numpy()])
    )

    Vrange_kaf[unit] = seasonal_ranges.quantile(0.95)

    print(
        unit.upper(),
        f"median = {seasonal_ranges.median():.2f} kaf,",
        f"95% = {Vrange_kaf[unit]:.2f} kaf",
    )
```
